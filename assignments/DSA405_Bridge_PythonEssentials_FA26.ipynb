## 2 — Lists and slicing

A list is ordered and zero-indexed: the first item is `[0]`. Slices are `[start:stop]`,
**including** start, **excluding** stop — so `[:3]` is the first three, `[-1]` is the last
item, and `[-3:]` is the last three.

```python
scores = [88.5, 92.0, 79.5, 95.0, 84.0, 91.5]

print("first:", scores[0], "| last:", scores[-1])
print("first three:", scores[:3])
print("last three :", scores[-3:])

scores.append(77.0)          # modifies in place, returns nothing
scores.sort()                # also in place
print("sorted:", scores)
print("how many:", len(scores))
```

## 3 — Dictionaries

A dict maps keys to values. This is the single most load-bearing structure in the course:
JSON files, API responses, and every row a scraper produces are dicts or lists of dicts.

The difference between `d[k]` and `d.get(k)` matters: the first **crashes** on a missing
key, the second returns `None` (or a default you choose). Both behaviours are useful — by
Week 10 you will be choosing between them deliberately.

```python
ratings = {"Port City Java": 92.5, "Los Lobos": 88.0}

print(ratings["Port City Java"])          # crashes if absent
print(ratings.get("Talley Market"))       # None if absent
print(ratings.get("Talley Market", 0))    # default if absent

ratings["Talley Market"] = 90.5           # add or overwrite
print(len(ratings), "entries:", list(ratings.keys()))
```

## 4 — Loops

`for x in things:` walks any collection. Two patterns cover most of data work: accumulate
(build a total or a list as you go) and filter-count (tally the items meeting a condition).

```python
all_scores = [88.5, 92.0, 79.5, 95.0]

total = 0
n_low = 0
for s in all_scores:
    total = total + s
    if s < 90:
        n_low = n_low + 1

print(f"mean by hand: {total / len(all_scores):.2f}")
print(f"{n_low} of {len(all_scores)} scores below 90")

# looping over a dict gives keys; .items() gives both
for name, r in {"Port City Java": 92.5, "Los Lobos": 88.0}.items():
    print(f"  {name}: {r}")
```

## 5 — Comprehensions

The catalog lists these as a prerequisite, and here is why: `[expression for x in things
if condition]` builds a list in one readable line, and it is the plain-Python shape of what
`pandas` does to whole columns. Read it left to right: *keep the expression, for each x,
where the condition holds.*

```python
raw = ["88.5", " 92.0", "79.5 ", "95.0"]

clean = [float(s) for s in raw]                 # transform every item
low   = [x for x in clean if x < 90]            # filter
names = {"Raleigh": 3, "Cary": 2, "Apex": 1}
big   = [city for city, n in names.items() if n >= 2]

print(clean)
print(low)
print(big)
```

## 6 — Functions

`def` packages steps under a name. The docstring (the string right under the `def` line)
says what the function expects — a habit this course grades from Week 9 onward, so start
now. Arguments can have defaults; `return` hands the value back.

```python
def mean(values):
    """Arithmetic mean of a non-empty list of numbers."""
    return sum(values) / len(values)

def share_below(values, cutoff=90):
    """Fraction of values strictly below cutoff."""
    hits = [v for v in values if v < cutoff]
    return len(hits) / len(values)

print(mean([2, 4, 9]))
print(share_below([88.5, 92.0, 79.5, 95.0]))          # default cutoff
print(share_below([88.5, 92.0, 79.5, 95.0], 95))      # your own cutoff
```

## 7 — None, truthiness, and the comparisons that lie

`None` is Python's "no value." It is not zero, not an empty string, and not `False` — and
the difference between `None` and `0` is, genuinely, a Week 3 exam question in every data
job on earth. Test for it with `is None`.

Empty things are "falsy": `[]`, `{}`, `""`, and `0` all count as `False` in an `if`. That
is convenient and occasionally dangerous — an `if score:` treats a legitimate score of 0
the same as a missing one.

```python
score = None
print(score is None)         # the right test
print(score == 0)            # False — None is not zero

for value in [0, "", [], None, 42, "0"]:
    print(f"{value!r:6} -> {'truthy' if value else 'falsy'}")

# note "0" (the string) is truthy. The string is not the number.
```

## 8 — Reading a traceback

Errors are not failures; they are the interpreter telling you exactly where to look. Read a
traceback **from the bottom up**: the last line names the exception and says what went
wrong; the lines above it point at the file and line. The cell below triggers an error on
purpose, catches it, and prints the parts.

```python
def total_seats(rows):
    total = 0
    for r in rows:
        total = total + r["seats"]
    return total

rows = [{"seats": 40}, {"seats": "55"}, {"seats": 32}]

try:
    total_seats(rows)
except TypeError as e:
    print("exception type   : TypeError")
    print("its message      :", e)
    print("what it means    : one of the seats values is a string, not a number.")
    print("where to look    : the line doing the +, and then at your DATA, not your code.")
```

## 9 — Putting it together: a table, before pandas

Raw data usually arrives as a list of dicts — one dict per row. Everything you just
practised composes into real questions. Do this once by hand, so that when `pandas` does
it in one line next week, you know exactly what that line is doing for you.

```python
inspections = [
    {"name": "Hargett Fish Camp",  "city": "Fuquay-Varina", "score": 86.7, "critical": 2},
    {"name": "Cabarrus Provisions","city": "Knightdale",    "score": 94.0, "critical": 0},
    {"name": "Jones Counter",      "city": "Knightdale",    "score": 95.2, "critical": 0},
    {"name": "Peace Smokehouse",   "city": "Holly Springs", "score": 100.0,"critical": 0},
]

scores = [r["score"] for r in inspections]
print(f"{len(inspections)} inspections, mean score {sum(scores)/len(scores):.1f}")

flagged = [r["name"] for r in inspections if r["critical"] > 0]
print("with critical violations:", flagged)

by_city = {}
for r in inspections:
    by_city[r["city"]] = by_city.get(r["city"], 0) + 1
print("inspections per city:", by_city)

# Next week, each of those is one line:
#   df.score.mean() · df[df.critical > 0] · df.city.value_counts()
```
